```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("2025.csv", low_memory=False)
```

```python
df = df[df["nom_region"] == 'Metropolitana de Santiago'].copy()
```

```python
# Convertir a numérico (por si están como texto en el CSV)
df['latitud'] = pd.to_numeric(df['latitud'], errors='coerce')
df['longitud'] = pd.to_numeric(df['longitud'], errors='coerce')

# 4. Obtener los pares únicos de latitud y longitud
coordenadas_unicas = df.dropna(subset=['latitud', 'longitud']).drop_duplicates(subset=['latitud', 'longitud'])

# 5. Plotear las bencineras
plt.figure(figsize=(10, 8))
plt.scatter(coordenadas_unicas['longitud'], coordenadas_unicas['latitud'],
            c='red', marker='o', alpha=0.5, s=20)

plt.title('Bencineras en la Región Metropolitana')
plt.xlabel('Longitud')
plt.ylabel('Latitud')
plt.grid(True)
plt.show()
```

```python
import os
import shutil

base_dir = r"datsets/precios bencina"

for item in os.listdir(base_dir):
    folder_path = os.path.join(base_dir, item)

    if os.path.isdir(folder_path) and item.endswith('.csv'):
        # 1. Rename the folder by removing '.csv'
        new_folder_name = item[:-4]
        new_folder_path = os.path.join(base_dir, new_folder_name)
        os.rename(folder_path, new_folder_path)

        # 2. Move the file
        file_path = os.path.join(new_folder_path, item)
        if os.path.isfile(file_path):
            shutil.move(file_path, base_dir)
            # 3. Remove the empty folder
            os.rmdir(new_folder_path)
            print(f"Moved {item} and removed the folder {new_folder_name}.")
```
